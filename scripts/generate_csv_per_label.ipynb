```python
import pandas as pd
import os
```

```python
INPUT_FILE  = r"C:\Users\hp\Desktop\stage etis\open_dataset\goodscents_jadbio_ready.csv"
OUTPUT_DIR  = r"C:\Users\hp\Desktop\stage etis\open_dataset\jadbio_csvs"
SEPARATOR   = ";" 


ALL_LABELS = [
    "floral",
    "fruity",
    "sweet",
    "woody",
    "green",
    "spicy",
    "animal_musk",
    "earthy",
    "citrus",
    "chemical",
    "gourmand",
    "powdery_amber",
]


```

```python
df = pd.read_csv(INPUT_FILE, sep=SEPARATOR)

non_feature_cols = ["SMILES"] + ALL_LABELS
descriptor_cols  = [c for c in df.columns if c not in non_feature_cols]
print(f"Descriptor columns found: {len(descriptor_cols)}")
print(f"Labels to process: {len(ALL_LABELS)}\n")
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"{'Label':<15} {'Rows':>6} {'Positive':>10} {'% Positive':>12}")
print("-" * 48)
 
for label in ALL_LABELS:
    if label not in df.columns:
        print(f"{label:<15}   column not found — skipping")
        continue
    subset = df[descriptor_cols + [label]].dropna(subset=[label])
 
    n_total    = len(subset)
    n_positive = int(subset[label].sum())
    pct        = 100 * n_positive / n_total

    out_path = os.path.join(OUTPUT_DIR, f"jadbio_{label}.csv")
    subset.to_csv(out_path, index=False, sep=SEPARATOR)
 
    print(f"{label:<15} {n_total:>6} {n_positive:>10} {pct:>11.1f}%")
```

```python
print(f"\nDone! {len(ALL_LABELS)} files saved to '{OUTPUT_DIR}/'")
```
