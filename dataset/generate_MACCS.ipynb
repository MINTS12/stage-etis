```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import MACCSkeys
```

```python
INPUT_FILE  =r"C:\Users\hp\Desktop\stage etis\dataset\7Q27.signatures_with_smiles.csv"
OUTPUT_FILE = r'C:\Users\hp\Desktop\stage etis\dataset\7Q27_signatures_with_maccs.csv'
 
EXCLUDED_ITEMS = {"Blanc_080426", "allumage_080426"}
```

```python
df= pd.read_csv(INPUT_FILE , sep=";")
df_clean=df[~df["item"].isin(EXCLUDED_ITEMS)].copy()
df_excluded=df[df["item"].isin(EXCLUDED_ITEMS)]
```

```python
def smiles_to_maccs(smiles) :
    # in the documentation they work with objects of the rdchem.Mol class so 5ss nconvertiwha lobjects dyal dik lclass
    mol = Chem.MolFromSmiles(smiles)

    fp = MACCSkeys.GenMACCSKeys(mol)
    return list(fp) 

unique_smiles = df_clean[["item", "SMILES"]].drop_duplicates(subset="SMILES")
 
print("Computing MACCS fingerprints for each unique molecule:\n")
maccs_data = []
 
for s in unique_smiles["SMILES"]:
    bits = smiles_to_maccs(s)
    if bits:
        # Create a dictionary with SMILES and bits 1-166
        entry = {"SMILES": s}
        entry.update({f"MACCS_{i:03d}": bits[i] for i in range(1, 167)})
        maccs_data.append(entry)

# 4. Merge and Save
maccs_df = pd.DataFrame(maccs_data)
df_final = df.merge(maccs_df, on="SMILES", how="inner")
df_final.to_csv("7Q27_signatures_with_maccs.csv", sep=";", index=False)

print(f"Done. Saved {len(df_final)} rows.")
 

```
